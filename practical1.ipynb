### 1. WAP to compute factorial of a given number

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        res = 1
        for i in range(1, n + 1):
            res *= i
        return res

# Test the function
num = 5
print(f"The factorial of {num} is {factorial(num)}")

num = 0
print(f"The factorial of {num} is {factorial(num)}")

num = 7
print(f"The factorial of {num} is {factorial(num)}")
```

### 2. WAP to display table for a given number

```python
def display_multiplication_table(num):
    print(f"Multiplication Table for {num}:")
    for i in range(1, 11):
        print(f"{num} x {i} = {num * i}")

# Test the function
display_multiplication_table(7)
print("\n")
display_multiplication_table(12)
```

### 3. WAP to identify the given number is odd or even

```python
def check_odd_even(num):
    if num % 2 == 0:
        print(f"{num} is an even number.")
    else:
        print(f"{num} is an odd number.")

# Test the function
check_odd_even(4)
check_odd_even(7)
check_odd_even(0)
```

### 4. WAP to identify the entered number is palindrome or not.

```python
def check_palindrome(num):
    # Convert number to string to easily reverse it
    s_num = str(num)
    reversed_s_num = s_num[::-1]

    if s_num == reversed_s_num:
        print(f"{num} is a palindrome.")
    else:
        print(f"{num} is not a palindrome.")

# Test the function
check_palindrome(121)
check_palindrome(123)
check_palindrome(1221)
check_palindrome(7)

# Also works for strings if input is handled as such
def check_string_palindrome(text):
    clean_text = text.lower().replace(" ", "") # Optional: make case-insensitive and ignore spaces
    if clean_text == clean_text[::-1]:
        print(f"'{text}' is a palindrome.")
    else:
        print(f"'{text}' is not a palindrome.")

check_string_palindrome("madam")
check_string_palindrome("A man a plan a canal Panama")
check_string_palindrome("hello")
```

### 5. Develop an Expert System that provides simple decision-making.

```python
def simple_expert_system():
    print("Welcome to the Simple Activity Advisor Expert System!")
    print("Please answer a few questions to get an activity suggestion.")

    while True:
        weather = input("What's the weather like today? (sunny/cloudy/rainy): ").lower()
        if weather in ['sunny', 'cloudy', 'rainy']:
            break
        else:
            print("Invalid weather input. Please choose from sunny, cloudy, or rainy.")

    while True:
        mood = input("How are you feeling today? (happy/neutral/sad): ").lower()
        if mood in ['happy', 'neutral', 'sad']:
            break
        else:
            print("Invalid mood input. Please choose from happy, neutral, or sad.")

    print("\n--- Decision Making ---")

    # Decision rules
    if weather == 'sunny':
        if mood == 'happy':
            print("Suggestion: Go for a picnic or play outdoor sports!")
        elif mood == 'neutral':
            print("Suggestion: Take a pleasant walk in the park.")
        elif mood == 'sad':
            print("Suggestion: Get some sun; it might lift your spirits! Maybe a gentle walk.")
    elif weather == 'cloudy':
        if mood == 'happy':
            print("Suggestion: Visit a museum or go to a coffee shop.")
        elif mood == 'neutral':
            print("Suggestion: Read a book or watch a movie.")
        elif mood == 'sad':
            print("Suggestion: Listen to some comforting music or try a new indoor hobby.")
    elif weather == 'rainy':
        if mood == 'happy':
            print("Suggestion: Have a cozy movie marathon or bake something delicious.")
        elif mood == 'neutral':
            print("Suggestion: Work on a puzzle or learn a new skill online.")
        elif mood == 'sad':
            print("Suggestion: Enjoy a warm beverage and reflect, or journal your thoughts.")
    else:
        print("Sorry, I couldn't make a suggestion based on the provided information.")

    print("\nThank you for using the Simple Activity Advisor!")

# Run the expert system
simple_expert_system()
```
